```python
import numpy as np
import pandas as pd
from scipy.stats import rankdata
from minepy import MINE  # Тот самый оригинальный импорт!

# =====================================================================
# 1. ГЕНЕРАЦИЯ СИНТЕТИКИ
# =====================================================================
def generate_synthetic_benchmarks(n_samples=100, noise_level=0.1):
    np.random.seed(42)
    data = {}
    x_lin = np.random.uniform(-1, 1, n_samples)
    data['Lin_X'] = x_lin
    data['Lin_Y'] = 1.5 * x_lin + np.random.normal(0, noise_level, n_samples)
    
    x_para = np.random.uniform(-1, 1, n_samples)
    data['Para_X'] = x_para
    data['Para_Y'] = 2.0 * (x_para ** 2) + np.random.normal(0, noise_level, n_samples)
    
    x_sine = np.random.uniform(-np.pi, np.pi, n_samples)
    data['Sine_X'] = x_sine
    data['Sine_Y'] = np.sin(2 * x_sine) + np.random.normal(0, noise_level, n_samples)
    
    theta = np.random.uniform(0, 2 * np.pi, n_samples)
    r = 1.0 + np.random.normal(0, noise_level * 0.5, n_samples)
    data['Circ_X'] = r * np.cos(theta)
    data['Circ_Y'] = r * np.sin(theta)
    
    data['Noise_1'] = np.random.normal(0, 1, n_samples)
    data['Noise_2'] = np.random.uniform(-1, 1, n_samples)
    return pd.DataFrame(data)

# =====================================================================
# 2. ФУНКЦИИ МЕР ЗАВИСИМОСТИ
# =====================================================================

def calculate_true_mic(x, y):
    """ Оригинальный расчет MIC через оптимизацию сетки (MINE suite) """
    # Параметры по умолчанию из оригинальной статьи Решефа (alpha=0.6, c=15)
    mine = MINE(alpha=0.6, c=15, est="mic_approx")
    mine.compute_score(x, y)
    return mine.mic()

def chatterjee_corr(x, y):
    n = len(x)
    idx = np.argsort(x)
    y_ordered = y[idx]
    temp = y_ordered.argsort()
    ranks = np.empty_like(temp)
    ranks[temp] = np.arange(1, n + 1)
    return 1 - (3 * np.sum(np.abs(np.diff(ranks)))) / (n**2 - 1)

def dist_corr(x, y):
    n = len(x)
    a = np.abs(x[:, None] - x[None, :])
    b = np.abs(y[:, None] - y[None, :])
    A = a - a.mean(axis=0, keepdims=True) - a.mean(axis=1, keepdims=True) + a.mean()
    B = b - b.mean(axis=0, keepdims=True) - b.mean(axis=1, keepdims=True) + b.mean()
    dcov2 = np.sum(A * B) / (n * n)
    dcov_xx = np.sum(A * A) / (n * n)
    dcov_yy = np.sum(B * B) / (n * n)
    return np.sqrt(dcov2 / np.sqrt(dcov_xx * dcov_yy)) if dcov_xx * dcov_yy > 0 else 0.0

def hoeffding_d(x, y):
    n = len(x)
    if n <= 4: return 0.0
    rx = rankdata(x); ry = rankdata(y)
    Q = np.zeros(n)
    for i in range(n):
        Q[i] = np.sum((rx < rx[i]) & (ry < ry[i]))
    D1 = np.sum((Q - 1) * (Q - 2))
    D2 = np.sum((rx - 1) * (rx - 2) * (ry - 1) * (ry - 2))
    D3 = np.sum((rx - 2) * (ry - 2) * (Q - 1))
    D = 30 * ((n - 2) * (n - 3) * D1 + D2 - 2 * (n - 2) * D3) / (n * (n - 1) * (n - 2) * (n - 3) * (n - 4))
    return max(0.0, D)

# =====================================================================
# 3. ПОСТРОЕНИЕ МАТРИЦ СМЕЖНОСТИ
# =====================================================================
df = generate_synthetic_benchmarks(n_samples=100)
cols = df.columns

adj_pearson = df.corr(method='pearson').abs()
adj_chatterjee = pd.DataFrame(index=cols, columns=cols, dtype=float)
adj_dcor = pd.DataFrame(index=cols, columns=cols, dtype=float)
adj_hoeffding = pd.DataFrame(index=cols, columns=cols, dtype=float)
adj_mic = pd.DataFrame(index=cols, columns=cols, dtype=float) # Матрица для НАСТОЯЩЕГО MIC

print("Вычисляем истинный MIC и остальные меры...")
for c1 in cols:
    for c2 in cols:
        if c1 == c2:
            adj_chatterjee.loc[c1, c2] = 1.0
            adj_dcor.loc[c1, c2] = 1.0
            adj_hoeffding.loc[c1, c2] = 1.0
            adj_mic.loc[c1, c2] = 1.0
        else:
            x_val, y_val = df[c1].values, df[c2].values
            adj_chatterjee.loc[c1, c2] = chatterjee_corr(x_val, y_val)
            adj_dcor.loc[c1, c2] = dist_corr(x_val, y_val)
            adj_hoeffding.loc[c1, c2] = hoeffding_d(x_val, y_val)
            adj_mic.loc[c1, c2] = calculate_true_mic(x_val, y_val)

print("--- 1. МАТРИЦА СМЕЖНОСТИ: МОДУЛЬ ПИРССОНА ---")
print(adj_pearson.round(3))

print("\n--- 2. МАТРИЦА СМЕЖНОСТИ: КОЭФФИЦИЕНТ ЧАТТЕРДЖИ ---")
print(adj_chatterjee.round(3))

print("\n--- 3. МАТРИЦА СМЕЖНОСТИ: ДИСТАНЦИОННАЯ КОРРЕЛЯЦИЯ ---")
print(adj_dcor.round(3))

print("\n--- 4. МАТРИЦА СМЕЖНОСТИ: МЕРА ХЕФДИНГА (D) ---")
print(adj_hoeffding.round(3))

print("\n--- 5. МАТРИЦА СМЕЖНОСТИ: ВЗАИМНАЯ ИНФОРМАЦИЯ (MI PROXY) ---")
print(adj_mic.round(3))
```

```python
import pandas as pd
import numpy as np

# 1. Загружаем файл (укажи правильный путь, если он лежит в папке)
df = pd.read_csv('Breast_GSE45827.csv.zip')

print(f"Исходный размер таблицы: {df.shape}")

# 2. Сохраняем клинические типы рака (они пригодятся для финальной красивой интерпретации)
tumor_types = df['type']

# 3. Отбираем только числовые столбцы (то есть чистую экспрессию генов)
gene_expression = df.select_dtypes(include=[np.number])

# 4. Считаем дисперсию (изменчивость) для каждого гена по всем 151 пациентам
variances = gene_expression.var(axis=0)

# 5. Выбираем топ-100 генов с самой высокой дисперсией
top_100_genes_list = variances.nlargest(100).index
filtered_matrix = gene_expression[top_100_genes_list]

print(f"Новый размер матрицы для расчета: {filtered_matrix.shape}")
print("Данные успешно сжаты до 100 самых активных генов!")
```

```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from scipy.stats import rankdata, pearsonr
from sklearn.feature_selection import mutual_info_regression
from minepy import MINE

# Переводим DataFrame в numpy-массив для вычислений
X_real = filtered_matrix.values
n_genes = X_real.shape[1]
gene_names = filtered_matrix.columns

# --- 1. ОПРЕДЕЛЕНИЕ ФУНКЦИЙ ДЛЯ МЕР ЗАВИСИМОСТИ ---

def chatterjee_coefficient(x, y):
    n = len(x)
    sort_idx = np.argsort(x)
    y_sorted = y[sort_idx]
    ranks = rankdata(y_sorted, method='ordinal')
    return 1 - (3 * np.sum(np.abs(np.diff(ranks)))) / (n**2 - 1)

def distance_correlation(x, y):
    n = len(x)
    X = np.abs(x[:, None] - x[None, :])
    Y = np.abs(y[:, None] - y[None, :])
    A = X - X.mean(axis=1, keepdims=True) - X.mean(axis=0, keepdims=True) + X.mean()
    B = Y - Y.mean(axis=1, keepdims=True) - Y.mean(axis=0, keepdims=True) + Y.mean()
    dcov2 = np.sum(A * B) / (n * n)
    dvarX2 = np.sum(A * A) / (n * n)
    dvarY2 = np.sum(B * B) / (n * n)
    if dvarX2 == 0 or dvarY2 == 0:
        return 0.0
    return np.sqrt(dcov2 / np.sqrt(dvarX2 * dvarY2))

def hoeffdings_d(x, y):
    """Коэффициент зависимости Хефдинга (Hoeffding's D)"""
    n = len(x)
    R = rankdata(x)
    S = rankdata(y)
    Q = np.zeros(n)
    for i in range(n):
        # Количество точек, где обе координаты строго меньше текущей
        Q[i] = np.sum((x < x[i]) & (y < y[i]))
    
    D1 = np.sum((Q - 1.0) * (Q - 2.0))
    D2 = np.sum((R - 1.0) * (R - 2.0) * (S - 1.0) * (S - 2.0))
    D3 = np.sum((R - 2.0) * (S - 2.0) * (Q - 1.0))
    
    # Итоговая формула для D
    D = 30.0 * ((n - 2.0) * (n - 3.0) * D1 + D2 - 2.0 * (n - 2.0) * D3) / (n * (n - 1.0) * (n - 2.0) * (n - 3.0) * (n - 4.0))
    return max(0.0, D) # Ограничиваем снизу нулем для корреляционного графа

def get_mic(x, y):
    """Maximal Information Coefficient (MIC)"""
    mine = MINE(alpha=0.6, c=15)
    mine.compute_score(x, y)
    return mine.mic()

# --- 2. ИНИЦИАЛИЗАЦИЯ МАТРИЦ СМЕЖНОСТИ ---
matrix_pearson = np.zeros((n_genes, n_genes))
matrix_chatterjee = np.zeros((n_genes, n_genes))
matrix_dcor = np.zeros((n_genes, n_genes))
matrix_mi = np.zeros((n_genes, n_genes))
matrix_hoeffding = np.zeros((n_genes, n_genes))
matrix_mic = np.zeros((n_genes, n_genes))

print("Запуск расчетов для 100 генов (около 5000 пар). Это займет пару минут...")

# --- 3. ВЫЧИСЛИТЕЛЬНЫЙ ЦИКЛ ---
for i in range(n_genes):
    for j in range(i, n_genes):
        x = X_real[:, i]
        y = X_real[:, j]
        
        if i == j:
            matrix_pearson[i, j] = 1.0
            matrix_chatterjee[i, j] = 1.0
            matrix_dcor[i, j] = 1.0
            matrix_hoeffding[i, j] = 1.0
            matrix_mic[i, j] = 1.0
            mi_val = mutual_info_regression(x.reshape(-1, 1), x)[0]
            matrix_mi[i, j] = mi_val
        else:
            # 1. Пирсон
            p_corr, _ = pearsonr(x, y)
            matrix_pearson[i, j] = matrix_pearson[j, i] = abs(p_corr)
            
            # 2. Чаттерджи
            chatt_ij = chatterjee_coefficient(x, y)
            chatt_ji = chatterjee_coefficient(y, x)
            matrix_chatterjee[i, j] = matrix_chatterjee[j, i] = (chatt_ij + chatt_ji) / 2
            
            # 3. Дистанционная корреляция
            dc = distance_correlation(x, y)
            matrix_dcor[i, j] = matrix_dcor[j, i] = dc
            
            # 4. Взаимная информация
            mi_val = mutual_info_regression(x.reshape(-1, 1), y)[0]
            matrix_mi[i, j] = matrix_mi[j, i] = max(0, mi_val)
            
            # 5. Хефдинг
            hoeff_val = hoeffdings_d(x, y)
            matrix_hoeffding[i, j] = matrix_hoeffding[j, i] = hoeff_val
            
            # 6. MIC (minepy)
            mic_val = get_mic(x, y)
            matrix_mic[i, j] = matrix_mic[j, i] = mic_val

print("Все матрицы успешно рассчитаны!")

# --- 4. ВИЗУАЛИЗАЦИЯ РЕЗУЛЬТАТОВ (Сетка 2x3) ---
fig, axes = plt.subplots(2, 3, figsize=(22, 14))

sns.heatmap(matrix_pearson, cmap='viridis', ax=axes[0, 0])
axes[0, 0].set_title("1. Модуль Пирсона")

sns.heatmap(matrix_chatterjee, cmap='plasma', ax=axes[0, 1])
axes[0, 1].set_title("2. Коэффициент Чаттерджи")

sns.heatmap(matrix_dcor, cmap='inferno', ax=axes[0, 2])
axes[0, 2].set_title("3. Дистанционная корреляция")

sns.heatmap(matrix_hoeffding, cmap='cividis', ax=axes[1, 0])
axes[1, 0].set_title("4. Коэффициент Хефдинга (D)")

sns.heatmap(matrix_mic, cmap='magma', ax=axes[1, 1])
axes[1, 1].set_title("5. MIC (Maximal Information Coefficient)")

sns.heatmap(matrix_mi, cmap='rocket', ax=axes[1, 2])
axes[1, 2].set_title("6. Взаимная информация (MI)")

plt.tight_layout()
plt.show()
```

```python
import networkx as nx
import numpy as np
import matplotlib.pyplot as plt

# Список наших матриц и их названий для цикла
matrices = {
    "Пирсон": matrix_pearson,
    "Чаттерджи": matrix_chatterjee,
    "dCor": matrix_dcor,
    "Хефдинг": matrix_hoeffding,
    "MIC": matrix_mic,
    "Взаимная информация": matrix_mi
}

# Задаем количество ребер, которое мы хотим оставить в каждом графе
N_EDGES_TO_KEEP = 100
n_nodes = 100

fig, axes = plt.subplots(2, 3, figsize=(22, 14))
axes = axes.flatten()

# Фиксируем расположение вершин (layout) на основе одной из матриц, 
# либо используем круговую разметку, чтобы четко видеть топологические различия
# Для наглядности распределения связей круговая разметка (circular_layout) подходит идеально
G_dummy = nx.complete_graph(n_nodes)
pos = nx.circular_layout(G_dummy)

for idx, (name, mat) in enumerate(matrices.items()):
    # Обнуляем диагональ, чтобы гены не связывались сами с собой
    mat_copy = mat.copy()
    np.fill_diagonal(mat_copy, 0)
    
    # Берем только верхний треугольник матрицы (так как граф ненаправленный)
    upper_tri_indices = np.triu_indices(n_nodes, k=1)
    edge_values = mat_copy[upper_tri_indices]
    
    # Находим пороговое значение, выше которого останется ровно N_EDGES_TO_KEEP ребер
    if len(edge_values) > N_EDGES_TO_KEEP:
        threshold = np.sort(edge_values)[-N_EDGES_TO_KEEP]
    else:
        threshold = 0
        
    # Бинаризируем матрицу смежности по найденному порогу
    adj_matrix = (mat_copy >= threshold).astype(int)
    
    # Строим граф из бинарной матрицы смежности
    G = nx.from_numpy_array(adj_matrix)
    
    # Считаем базовые топологические метрики для диплома
    isolates = list(nx.isolates(G))
    n_connected_components = nx.number_connected_components(G)
    
    # Отрисовка графа
    ax = axes[idx]
    
    # Рисуем изолированные вершины серым, а связанные — синим
    node_colors = ['#ced4da' if node in isolates else '#1d3557' for node in G.nodes()]
    node_sizes = [20 if node in isolates else 50 for node in G.nodes()]
    
    nx.draw_networkx_nodes(G, pos, node_size=node_sizes, node_color=node_colors, ax=ax)
    nx.draw_networkx_edges(G, pos, alpha=0.4, edge_color='#e63946', width=1.2, ax=ax)
    
    ax.set_title(f"{idx+1}. {name}\nКомпонент связности: {n_connected_components}\nИзолированных генов: {len(isolates)}", 
                 fontsize=14, fontweight='bold')
    ax.axis('off')

plt.tight_layout()
plt.show()
```
